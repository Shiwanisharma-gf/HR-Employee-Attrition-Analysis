```python
result=gh.groupby('gender')['years_at_company'].mean()
plt.figure(figsize=(8,6))
plt.barh(result.index,result.values,label='satying in company')
plt.xlabel('gender')
plt.ylabel(' total_years_at_company')
plt.legend()
plt.title('years at company by gender')
plt.show()
```

```python
# 1. Groupby calculation
scores_summary = gh.groupby('attrition')[['burnout_score', 'work_life_balance_score']].mean()

# 2. Direct DataFrame plotting (ye automatically dono columns ke side-by-side bars bana dega)
scores_summary.plot(kind='bar', figsize=(8, 6), color=['red', 'green'])

# 3. Titles & Labels
plt.title('Burnout & Work-Life Balance Score by Attrition')
plt.xlabel('Attrition (Left Company)')
plt.ylabel('Average Score (out of 10)')
plt.xticks(rotation=0)
plt.legend(['Burnout Score', 'Work-Life Balance Score'])
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```

```python
ai_risk_summary=gh.groupby('uses_ai_tools_at_work')['perceived_ai_job_risk'].mean()
plt.figure(figsize=(8,6))
ai_risk_summary.plot(kind='bar',color=['blue','black'])
plt.title('Average Perceived AI Job Risk by AI Tool Usage')
plt.xlabel('Uses AI Tools at Work')
plt.ylabel('Average Perceived AI Job Risk Score')
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()


```

```python
avg_income = gh.groupby('attrition')['monthly_income'].mean()

plt.figure(figsize=(6, 6))
plt.pie(avg_income, labels=avg_income.index, autopct="%1.1f%%", colors=['#2ecc71', '#e74c3c'], startangle=90)
plt.title('Average Monthly Income Share by Attrition')
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
### 2. Average Monthly Income & Job Level Comparison
# Average Monthly Income
gh.groupby('attrition')['monthly_income'].mean().plot(kind='bar', ax=axes[0], color='#3498db', edgecolor='black')
axes[0].set_title('Average Monthly Income by Attrition')
axes[0].set_xlabel('Attrition')
axes[0].set_ylabel('Monthly Income')
axes[0].tick_params(axis='x', rotation=0)

# Average Job Level
gh.groupby('attrition')['job_level'].mean().plot(kind='bar', ax=axes[1], color='#e67e22', edgecolor='black')
axes[1].set_title('Average Job Level by Attrition')
axes[1].set_xlabel('Attrition')
axes[1].set_ylabel('Job Level (1 to 5)')
axes[1].tick_params(axis='x', rotation=0)

plt.tight_layout()
plt.show()
```

```python
# ### 📌 Key Takeaways
# - Employees leaving the company have lower average monthly income and job levels.
# - Compensation and hierarchy play a visible role in employee retention.
```
